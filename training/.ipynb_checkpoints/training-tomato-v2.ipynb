```python
# ============================================================
# CELL 2: Imports
# ============================================================
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, Model
from tensorflow.keras.applications import EfficientNetV2S, ConvNeXtSmall
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
import matplotlib.pyplot as plt
import numpy as np
import cv2
import os
import json
from sklearn.model_selection import StratifiedKFold
from sklearn.utils.class_weight import compute_class_weight
from collections import Counter
import warnings
warnings.filterwarnings("ignore")

# MLflow for experiment tracking
import mlflow
import mlflow.keras

print(f"TensorFlow version: {tf.__version__}")
print(f"GPU Available: {tf.config.list_physical_devices('GPU')}")
```

```python
# ============================================================
# CELL 3: Configuration
# ============================================================
class Config:
    # Data
    DATA_DIR = "../Tomato"
    IMAGE_SIZE = 224  # EfficientNet/ConvNeXt optimal size
    BATCH_SIZE = 32
    CHANNELS = 3
    
    # Training
    EPOCHS = 50
    LEARNING_RATE = 1e-4
    WARMUP_EPOCHS = 5
    
    # Model - Choose backbone: "efficientnetv2" or "convnext"
    BACKBONE = "efficientnetv2"  # Modern CNN with SOTA performance
    DROPOUT_RATE = 0.3
    MC_DROPOUT_SAMPLES = 10  # For uncertainty estimation
    
    # Augmentation
    USE_MIXUP = True
    USE_CUTMIX = True
    MIXUP_ALPHA = 0.2
    CUTMIX_ALPHA = 1.0
    
    # Validation
    K_FOLDS = 5
    USE_KFOLD = False  # Set True for full k-fold training
    
    # Calibration
    CALIBRATION_TEMPERATURE = 1.5  # Will be optimized
    CONFIDENCE_THRESHOLD = 0.7  # Below this -> "Uncertain"
    
    # Paths
    SAVED_MODELS_DIR = "../saved_models_v2"
    MLFLOW_TRACKING_URI = "mlruns"
    
config = Config()
os.makedirs(config.SAVED_MODELS_DIR, exist_ok=True)

print(f"🔧 Configuration:")
print(f"   Backbone: {config.BACKBONE}")
print(f"   Image Size: {config.IMAGE_SIZE}x{config.IMAGE_SIZE}")
print(f"   Batch Size: {config.BATCH_SIZE}")
```

```python
# ============================================================
# CELL 4: Load Dataset
# ============================================================
dataset = tf.keras.preprocessing.image_dataset_from_directory(
    config.DATA_DIR,
    seed=42,
    shuffle=True,
    image_size=(config.IMAGE_SIZE, config.IMAGE_SIZE),
    batch_size=config.BATCH_SIZE,
    label_mode='int'
)

class_names = dataset.class_names
num_classes = len(class_names)

print(f"Classes ({num_classes}): {class_names}")
print(f"Total batches: {len(dataset)}")
```

```python
# ============================================================
# CELL 5: Compute Class Weights for Imbalanced Data
# ============================================================
# Extract all labels for class weight computation
all_labels = []
for _, labels in dataset:
    all_labels.extend(labels.numpy())

all_labels = np.array(all_labels)
class_counts = Counter(all_labels)
print("Class distribution:")
for idx, count in sorted(class_counts.items()):
    print(f"  {class_names[idx]}: {count} samples")

# Compute class weights
class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(all_labels),
    y=all_labels
)
class_weight_dict = dict(enumerate(class_weights))
print(f"\nClass weights: {class_weight_dict}")
```

```python
# ============================================================
# CELL 6: Data Splitting with Stratification
# ============================================================
def get_dataset_partitions(ds, train_split=0.8, val_split=0.1, test_split=0.1, shuffle=True):
    assert abs(train_split + val_split + test_split - 1.0) < 1e-5
    
    ds_size = len(ds)
    if shuffle:
        ds = ds.shuffle(buffer_size=1000, seed=42)
    
    train_size = int(train_split * ds_size)
    val_size = int(val_split * ds_size)
    
    train_ds = ds.take(train_size)
    val_ds = ds.skip(train_size).take(val_size)
    test_ds = ds.skip(train_size + val_size)
    
    return train_ds, val_ds, test_ds

train_ds, val_ds, test_ds = get_dataset_partitions(dataset)
print(f"Train batches: {len(train_ds)}, Val batches: {len(val_ds)}, Test batches: {len(test_ds)}")
```

```python
# ============================================================
# CELL 7: Advanced Augmentation Layer (RandAugment-style)
# ============================================================
class AdvancedAugmentation(layers.Layer):
    """Advanced augmentation with RandAugment-style transformations"""
    
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.augment = keras.Sequential([
            layers.RandomFlip("horizontal_and_vertical"),
            layers.RandomRotation(0.3),
            layers.RandomZoom(0.2),
            layers.RandomTranslation(0.1, 0.1),
            layers.RandomContrast(0.2),
            layers.RandomBrightness(0.2),
        ])
    
    def call(self, images, training=True):
        if training:
            return self.augment(images)
        return images

augmentation_layer = AdvancedAugmentation()
```

```python
# ============================================================
# CELL 8: MixUp and CutMix Implementation
# ============================================================
def mixup(images, labels, alpha=0.2):
    """MixUp augmentation"""
    batch_size = tf.shape(images)[0]
    
    # Sample lambda from Beta distribution
    lam = tf.random.uniform([], 0, alpha)
    
    # Shuffle indices
    indices = tf.random.shuffle(tf.range(batch_size))
    shuffled_images = tf.gather(images, indices)
    shuffled_labels = tf.gather(labels, indices)
    
    # Mix
    mixed_images = lam * images + (1 - lam) * shuffled_images
    mixed_labels = lam * tf.cast(labels, tf.float32) + (1 - lam) * tf.cast(shuffled_labels, tf.float32)
    
    return mixed_images, mixed_labels


def cutmix(images, labels, alpha=1.0):
    """CutMix augmentation"""
    batch_size = tf.shape(images)[0]
    img_h = tf.shape(images)[1]
    img_w = tf.shape(images)[2]
    
    # Sample lambda from Beta distribution
    lam = tf.random.uniform([], 0.3, 0.7)
    
    # Shuffle indices
    indices = tf.random.shuffle(tf.range(batch_size))
    shuffled_images = tf.gather(images, indices)
    shuffled_labels = tf.gather(labels, indices)
    
    # Get random box
    cut_ratio = tf.sqrt(1 - lam)
    cut_h = tf.cast(tf.cast(img_h, tf.float32) * cut_ratio, tf.int32)
    cut_w = tf.cast(tf.cast(img_w, tf.float32) * cut_ratio, tf.int32)
    
    cx = tf.random.uniform([], 0, img_w, dtype=tf.int32)
    cy = tf.random.uniform([], 0, img_h, dtype=tf.int32)
    
    x1 = tf.clip_by_value(cx - cut_w // 2, 0, img_w)
    x2 = tf.clip_by_value(cx + cut_w // 2, 0, img_w)
    y1 = tf.clip_by_value(cy - cut_h // 2, 0, img_h)
    y2 = tf.clip_by_value(cy + cut_h // 2, 0, img_h)
    
    # Create mask
    mask = tf.ones((img_h, img_w), dtype=tf.float32)
    padding = [[y1, img_h - y2], [x1, img_w - x2]]
    cutout = tf.zeros((y2 - y1, x2 - x1), dtype=tf.float32)
    mask = tf.tensor_scatter_nd_update(
        mask,
        tf.reshape(tf.stack(tf.meshgrid(tf.range(y1, y2), tf.range(x1, x2), indexing='ij'), axis=-1), [-1, 2]),
        tf.zeros([(y2 - y1) * (x2 - x1)], dtype=tf.float32)
    )
    mask = tf.reshape(mask, [1, img_h, img_w, 1])
    
    # Apply CutMix
    mixed_images = images * mask + shuffled_images * (1 - mask)
    
    # Adjust lambda based on actual cutout area
    adjusted_lam = 1 - tf.cast((x2 - x1) * (y2 - y1), tf.float32) / tf.cast(img_h * img_w, tf.float32)
    mixed_labels = adjusted_lam * tf.cast(labels, tf.float32) + (1 - adjusted_lam) * tf.cast(shuffled_labels, tf.float32)
    
    return mixed_images, mixed_labels


def apply_augmentation(images, labels):
    """Apply augmentation with probability"""
    images = augmentation_layer(images, training=True)
    
    # Randomly choose between MixUp, CutMix, or neither
    choice = tf.random.uniform([])
    
    if config.USE_MIXUP and choice < 0.33:
        images, labels = mixup(images, labels, config.MIXUP_ALPHA)
    elif config.USE_CUTMIX and choice < 0.66:
        images, labels = cutmix(images, labels, config.CUTMIX_ALPHA)
    else:
        labels = tf.cast(labels, tf.float32)
    
    return images, labels

print("✅ MixUp and CutMix implemented")
```

```python
# ============================================================
# CELL 9: Focal Loss for Class Imbalance
# ============================================================
class FocalLoss(keras.losses.Loss):
    """Focal Loss for handling class imbalance
    
    FL(p_t) = -alpha_t * (1 - p_t)^gamma * log(p_t)
    
    Args:
        gamma: Focusing parameter (default=2.0)
        alpha: Class weight balancing factor
    """
    
    def __init__(self, gamma=2.0, alpha=0.25, num_classes=None, **kwargs):
        super().__init__(**kwargs)
        self.gamma = gamma
        self.alpha = alpha
        self.num_classes = num_classes
    
    def call(self, y_true, y_pred):
        # Clip predictions to prevent log(0)
        y_pred = tf.clip_by_value(y_pred, 1e-7, 1 - 1e-7)
        
        # Handle soft labels (from MixUp/CutMix)
        if len(y_true.shape) == 1 or y_true.shape[-1] == 1:
            # Convert integer labels to one-hot
            y_true = tf.one_hot(tf.cast(y_true, tf.int32), self.num_classes)
            y_true = tf.squeeze(y_true)
        
        # Focal loss computation
        cross_entropy = -y_true * tf.math.log(y_pred)
        focal_weight = self.alpha * tf.pow(1 - y_pred, self.gamma)
        focal_loss = focal_weight * cross_entropy
        
        return tf.reduce_mean(tf.reduce_sum(focal_loss, axis=-1))
    
    def get_config(self):
        config = super().get_config()
        config.update({
            'gamma': self.gamma,
            'alpha': self.alpha,
            'num_classes': self.num_classes
        })
        return config

print("✅ Focal Loss implemented")
```

```python
# ============================================================
# CELL 10: Build Modern Model with Transfer Learning
# ============================================================
def build_model(num_classes, backbone="efficientnetv2", dropout_rate=0.3):
    """Build model with transfer learning backbone
    
    Args:
        num_classes: Number of output classes
        backbone: 'efficientnetv2' or 'convnext'
        dropout_rate: Dropout rate for MC Dropout
    """
    inputs = layers.Input(shape=(config.IMAGE_SIZE, config.IMAGE_SIZE, config.CHANNELS))
    
    # Select backbone
    if backbone == "efficientnetv2":
        print("🏗️ Building EfficientNetV2-S model...")
        print("   Architecture: Modern efficient CNN (2021)")
        print("   Pre-trained on: ImageNet")
        base_model = EfficientNetV2S(
            include_top=False,
            weights='imagenet',
            input_tensor=inputs,
            pooling='avg'
        )
        preprocess = tf.keras.applications.efficientnet_v2.preprocess_input
    elif backbone == "convnext":
        print("🏗️ Building ConvNeXt-Small model...")
        print("   Architecture: CNN with Transformer-inspired design (2022)")
        print("   Pre-trained on: ImageNet")
        base_model = ConvNeXtSmall(
            include_top=False,
            weights='imagenet',
            input_tensor=inputs,
            pooling='avg'
        )
        preprocess = tf.keras.applications.convnext.preprocess_input
    else:
        raise ValueError(f"Unknown backbone: {backbone}")
    
    # Freeze base model initially
    base_model.trainable = False
    
    # Build model
    x = preprocess(inputs)
    x = base_model(x, training=False)
    
    # Classification head with MC Dropout
    x = layers.Dense(256, activation='relu')(x)
    x = layers.Dropout(dropout_rate)(x, training=True)  # MC Dropout: always on
    x = layers.Dense(128, activation='relu')(x)
    x = layers.Dropout(dropout_rate)(x, training=True)  # MC Dropout: always on
    
    outputs = layers.Dense(num_classes, activation='softmax', name='predictions')(x)
    
    model = Model(inputs, outputs, name=f"{backbone}_classifier")
    
    return model, base_model


# Build model
model, base_model = build_model(
    num_classes=num_classes,
    backbone=config.BACKBONE,
    dropout_rate=config.DROPOUT_RATE
)

model.summary()
```

```python
# ============================================================
# CELL 11: Custom Metric for Soft Labels
# ============================================================
class SoftAccuracy(keras.metrics.Metric):
    """Accuracy metric that handles soft labels from MixUp/CutMix"""
    
    def __init__(self, name='accuracy', **kwargs):
        super().__init__(name=name, **kwargs)
        self.correct = self.add_weight(name='correct', initializer='zeros')
        self.total = self.add_weight(name='total', initializer='zeros')
    
    def update_state(self, y_true, y_pred, sample_weight=None):
        y_pred_class = tf.argmax(y_pred, axis=-1)
        
        # Handle soft labels
        if len(y_true.shape) > 1 and y_true.shape[-1] > 1:
            y_true_class = tf.argmax(y_true, axis=-1)
        else:
            y_true_class = tf.cast(tf.round(y_true), tf.int64)
            if len(y_true_class.shape) > 1:
                y_true_class = tf.squeeze(y_true_class, axis=-1)
        
        correct = tf.cast(tf.equal(y_pred_class, y_true_class), tf.float32)
        self.correct.assign_add(tf.reduce_sum(correct))
        self.total.assign_add(tf.cast(tf.shape(y_true)[0], tf.float32))
    
    def result(self):
        return self.correct / (self.total + 1e-7)
    
    def reset_state(self):
        self.correct.assign(0)
        self.total.assign(0)

print("✅ Soft label accuracy metric implemented")
```

```python
# ============================================================
# CELL 12: Learning Rate Schedule with Warmup
# ============================================================
class WarmupCosineDecay(keras.optimizers.schedules.LearningRateSchedule):
    """Learning rate schedule with warmup and cosine decay"""
    
    def __init__(self, base_lr, warmup_epochs, total_epochs, steps_per_epoch):
        super().__init__()
        self.base_lr = base_lr
        self.warmup_steps = warmup_epochs * steps_per_epoch
        self.total_steps = total_epochs * steps_per_epoch
    
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        
        # Warmup phase
        warmup_lr = self.base_lr * (step / self.warmup_steps)
        
        # Cosine decay phase
        decay_steps = self.total_steps - self.warmup_steps
        cosine_decay = 0.5 * (1 + tf.cos(np.pi * (step - self.warmup_steps) / decay_steps))
        decay_lr = self.base_lr * cosine_decay
        
        return tf.where(step < self.warmup_steps, warmup_lr, decay_lr)
    
    def get_config(self):
        return {
            'base_lr': self.base_lr,
            'warmup_steps': self.warmup_steps,
            'total_steps': self.total_steps
        }

# Create schedule
steps_per_epoch = len(train_ds)
lr_schedule = WarmupCosineDecay(
    base_lr=config.LEARNING_RATE,
    warmup_epochs=config.WARMUP_EPOCHS,
    total_epochs=config.EPOCHS,
    steps_per_epoch=steps_per_epoch
)

print("✅ Warmup + Cosine decay learning rate schedule created")
```

```python
# ============================================================
# CELL 13: Compile Model
# ============================================================
# Compile with Focal Loss
model.compile(
    optimizer=keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=1e-5),
    loss=FocalLoss(gamma=2.0, alpha=0.25, num_classes=num_classes),
    metrics=[SoftAccuracy()]
)

print("✅ Model compiled with Focal Loss and AdamW optimizer")
```

```python
# ============================================================
# CELL 14: Prepare Data Pipelines
# ============================================================
AUTOTUNE = tf.data.AUTOTUNE

# Training pipeline with augmentation
train_ds_aug = train_ds.map(
    apply_augmentation,
    num_parallel_calls=AUTOTUNE
).prefetch(AUTOTUNE)

# Validation/Test pipeline (no augmentation, convert labels to float)
val_ds_prep = val_ds.map(
    lambda x, y: (x, tf.cast(y, tf.float32)),
    num_parallel_calls=AUTOTUNE
).prefetch(AUTOTUNE)

test_ds_prep = test_ds.map(
    lambda x, y: (x, tf.cast(y, tf.float32)),
    num_parallel_calls=AUTOTUNE
).prefetch(AUTOTUNE)

print("✅ Data pipelines ready")
```

```python
# ============================================================
# CELL 15: MLflow Setup and Callbacks
# ============================================================
mlflow.set_tracking_uri(config.MLFLOW_TRACKING_URI)
mlflow.set_experiment("tomato-disease-v2")

callbacks = [
    EarlyStopping(
        monitor='val_loss',
        patience=10,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=5,
        min_lr=1e-7,
        verbose=1
    ),
    ModelCheckpoint(
        filepath=os.path.join(config.SAVED_MODELS_DIR, 'best_model.keras'),
        monitor='val_accuracy',
        mode='max',  # Maximize accuracy
        save_best_only=True,
        verbose=1
    )
]

print("✅ MLflow and callbacks configured")
```

```python
# ============================================================
# CELL 16: Train Model (Phase 1: Frozen Backbone)
# ============================================================
with mlflow.start_run(run_name="efficientnetv2_phase1_frozen"):
    # Log parameters
    mlflow.log_params({
        "backbone": config.BACKBONE,
        "image_size": config.IMAGE_SIZE,
        "batch_size": config.BATCH_SIZE,
        "learning_rate": config.LEARNING_RATE,
        "dropout_rate": config.DROPOUT_RATE,
        "use_mixup": config.USE_MIXUP,
        "use_cutmix": config.USE_CUTMIX,
        "focal_loss_gamma": 2.0,
        "phase": "frozen_backbone"
    })
    
    print("🚀 Phase 1: Training with frozen backbone...")
    history_phase1 = model.fit(
        train_ds_aug,
        validation_data=val_ds_prep,
        epochs=15,  # Initial phase with frozen backbone
        callbacks=callbacks,
        class_weight=class_weight_dict,
        verbose=1
    )
    
    # Log metrics
    mlflow.log_metric("phase1_val_accuracy", max(history_phase1.history['val_accuracy']))
```

```python
# ============================================================
# CELL 17: Fine-tune Model (Phase 2: Unfreeze Backbone)
# ============================================================
# Unfreeze the backbone for fine-tuning
base_model.trainable = True

# Fine-tune from a specific layer (keep early layers frozen)
for layer in base_model.layers[:-30]:
    layer.trainable = False

# Recompile with lower learning rate
model.compile(
    optimizer=keras.optimizers.AdamW(learning_rate=1e-5, weight_decay=1e-5),
    loss=FocalLoss(gamma=2.0, alpha=0.25, num_classes=num_classes),
    metrics=[SoftAccuracy()]
)

print(f"Trainable layers: {len([l for l in model.layers if l.trainable])}")

with mlflow.start_run(run_name="efficientnetv2_phase2_finetune"):
    mlflow.log_params({
        "phase": "fine_tuning",
        "learning_rate": 1e-5
    })
    
    print("🚀 Phase 2: Fine-tuning backbone...")
    history_phase2 = model.fit(
        train_ds_aug,
        validation_data=val_ds_prep,
        epochs=config.EPOCHS,
        initial_epoch=10,
        callbacks=callbacks,
        class_weight=class_weight_dict,
        verbose=1
    )
    
    mlflow.log_metric("final_val_accuracy", max(history_phase2.history['val_accuracy']))
```

```python
# ============================================================
# CELL 18: Combine Training History
# ============================================================
# Combine histories from both phases
history = {
    'loss': history_phase1.history['loss'] + history_phase2.history['loss'],
    'accuracy': history_phase1.history['accuracy'] + history_phase2.history['accuracy'],
    'val_loss': history_phase1.history['val_loss'] + history_phase2.history['val_loss'],
    'val_accuracy': history_phase1.history['val_accuracy'] + history_phase2.history['val_accuracy']
}

print(f"Total epochs trained: {len(history['loss'])}")
```

```python
# ============================================================
# CELL 20: Temperature Scaling for Calibration
# ============================================================
class TemperatureScaling:
    """Temperature scaling for model calibration
    
    Calibrates model confidence to match actual accuracy.
    """
    
    def __init__(self, model):
        self.model = model
        self.temperature = 1.0
    
    def find_optimal_temperature(self, val_ds, num_bins=15):
        """Find optimal temperature using validation set"""
        # Get predictions and labels
        all_logits = []
        all_labels = []
        
        for images, labels in val_ds:
            # Get pre-softmax logits (need to modify model temporarily)
            predictions = self.model(images, training=False)
            all_logits.append(predictions.numpy())
            all_labels.append(labels.numpy())
        
        all_logits = np.concatenate(all_logits, axis=0)
        all_labels = np.concatenate(all_labels, axis=0).astype(int)
        
        # Grid search for optimal temperature
        best_ece = float('inf')
        best_temp = 1.0
        
        for temp in np.linspace(0.5, 3.0, 50):
            scaled_probs = self._apply_temperature(all_logits, temp)
            ece = self._compute_ece(scaled_probs, all_labels, num_bins)
            
            if ece < best_ece:
                best_ece = ece
                best_temp = temp
        
        self.temperature = best_temp
        print(f"Optimal temperature: {self.temperature:.3f}")
        print(f"ECE after calibration: {best_ece:.4f}")
        return best_temp
    
    def _apply_temperature(self, logits, temperature):
        """Apply temperature scaling"""
        # Since we have softmax outputs, convert back to logits approx
        scaled = np.log(logits + 1e-7) / temperature
        exp_scaled = np.exp(scaled - np.max(scaled, axis=1, keepdims=True))
        return exp_scaled / exp_scaled.sum(axis=1, keepdims=True)
    
    def _compute_ece(self, probs, labels, num_bins=15):
        """Compute Expected Calibration Error"""
        confidences = np.max(probs, axis=1)
        predictions = np.argmax(probs, axis=1)
        accuracies = (predictions == labels).astype(float)
        
        bin_boundaries = np.linspace(0, 1, num_bins + 1)
        ece = 0.0
        
        for i in range(num_bins):
            mask = (confidences > bin_boundaries[i]) & (confidences <= bin_boundaries[i + 1])
            if mask.sum() > 0:
                bin_accuracy = accuracies[mask].mean()
                bin_confidence = confidences[mask].mean()
                ece += mask.sum() * abs(bin_accuracy - bin_confidence)
        
        return ece / len(labels)
    
    def predict_calibrated(self, images):
        """Get calibrated predictions"""
        raw_probs = self.model(images, training=False).numpy()
        return self._apply_temperature(raw_probs, self.temperature)

# Calibrate the model
calibrator = TemperatureScaling(model)
optimal_temp = calibrator.find_optimal_temperature(val_ds_prep)
```

```python
# ============================================================
# CELL 21: MC Dropout for Uncertainty Estimation
# ============================================================
def predict_with_uncertainty(model, images, n_samples=10):
    """Monte Carlo Dropout for uncertainty estimation
    
    Performs multiple forward passes with dropout enabled
    to estimate prediction uncertainty.
    
    Returns:
        mean_prediction: Average prediction across samples
        uncertainty: Standard deviation (epistemic uncertainty)
    """
    predictions = []
    
    for _ in range(n_samples):
        # training=True keeps dropout active
        pred = model(images, training=True)
        predictions.append(pred.numpy())
    
    predictions = np.array(predictions)  # Shape: (n_samples, batch, classes)
    
    mean_prediction = predictions.mean(axis=0)
    uncertainty = predictions.std(axis=0)
    
    # Predictive entropy (total uncertainty)
    predictive_entropy = -np.sum(mean_prediction * np.log(mean_prediction + 1e-7), axis=-1)
    
    # Mean class uncertainty
    mean_uncertainty = uncertainty.mean(axis=-1)
    
    return mean_prediction, mean_uncertainty, predictive_entropy

print("✅ MC Dropout uncertainty estimation implemented")
```

```python
# ============================================================
# CELL 22: Grad-CAM Implementation
# ============================================================
class GradCAM:
    """Grad-CAM visualization for model explainability"""
    
    def __init__(self, model, layer_name=None):
        self.model = model
        self.target_layer = None
        
        # Find last conv layer if not specified
        if layer_name is None:
            self.target_layer = self._find_target_layer(model)
        else:
            self.target_layer = model.get_layer(layer_name)
        
        if self.target_layer is None:
            raise ValueError("Could not find a suitable convolutional layer for Grad-CAM")
        
        print(f"Using layer '{self.target_layer.name}' for Grad-CAM")
        
        # Create gradient model using the actual layer output
        self.grad_model = Model(
            inputs=model.input,
            outputs=[self.target_layer.output, model.output]
        )
    
    def _find_target_layer(self, model):
        """Find the best layer for Grad-CAM visualization"""
        # Search through all layers including nested ones
        def search_layers(layers):
            candidates = []
            for layer in layers:
                # Check if this layer has 4D output (conv layer)
                try:
                    output_shape = layer.output.shape
                    if len(output_shape) == 4:
                        candidates.append(layer)
                except (AttributeError, ValueError, RuntimeError):
                    pass
                
                # Recursively search nested models
                if hasattr(layer, 'layers') and len(layer.layers) > 0:
                    nested_candidates = search_layers(layer.layers)
                    candidates.extend(nested_candidates)
            
            return candidates
        
        candidates = search_layers(model.layers)
        
        if candidates:
            # Return the last (deepest) conv layer found
            return candidates[-1]
        
        return None
    
    def compute_heatmap(self, image, class_idx=None, eps=1e-8):
        """Compute Grad-CAM heatmap for an image
        
        Args:
            image: Input image (H, W, C) or (1, H, W, C)
            class_idx: Target class index (None = predicted class)
        
        Returns:
            heatmap: Normalized heatmap (H, W)
        """
        if len(image.shape) == 3:
            image = np.expand_dims(image, 0)
        
        image = tf.cast(image, tf.float32)
        
        with tf.GradientTape() as tape:
            conv_outputs, predictions = self.grad_model(image)
            
            if class_idx is None:
                class_idx = tf.argmax(predictions[0])
            
            class_output = predictions[:, class_idx]
        
        # Compute gradients
        grads = tape.gradient(class_output, conv_outputs)
        
        # Handle case where gradients might be None
        if grads is None:
            print("Warning: Gradients are None, returning uniform heatmap")
            return np.ones((7, 7), dtype=np.float32) * 0.5
        
        # Global average pooling of gradients
        pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
        
        # Weight feature maps by gradients
        conv_outputs = conv_outputs[0]
        heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
        heatmap = tf.squeeze(heatmap)
        
        # Normalize heatmap
        heatmap = tf.maximum(heatmap, 0) / (tf.reduce_max(heatmap) + eps)
        
        return heatmap.numpy()
    
    def overlay_heatmap(self, image, heatmap, alpha=0.4, colormap=cv2.COLORMAP_JET):
        """Overlay heatmap on original image"""
        # Resize heatmap to image size
        heatmap_resized = cv2.resize(heatmap, (image.shape[1], image.shape[0]))
        
        # Convert to colormap
        heatmap_colored = cv2.applyColorMap(
            np.uint8(255 * heatmap_resized), colormap
        )
        heatmap_colored = cv2.cvtColor(heatmap_colored, cv2.COLOR_BGR2RGB)
        
        # Overlay
        if image.max() <= 1.0:
            image = (image * 255).astype(np.uint8)
        
        overlaid = cv2.addWeighted(image, 1 - alpha, heatmap_colored, alpha, 0)
        
        return overlaid

# Initialize Grad-CAM
gradcam = GradCAM(model)
```

```python
# ============================================================
# CELL 23: Comprehensive Prediction Function
# ============================================================
def predict_with_explanation(model, image, calibrator, gradcam, 
                              class_names, n_mc_samples=10,
                              confidence_threshold=0.7):
    """Complete prediction with uncertainty, calibration, and explanation
    
    Returns:
        result: Dictionary with prediction details
    """
    if len(image.shape) == 3:
        image_batch = np.expand_dims(image, 0)
    else:
        image_batch = image
    
    # MC Dropout predictions
    mean_pred, uncertainty, entropy = predict_with_uncertainty(
        model, image_batch, n_samples=n_mc_samples
    )
    
    # Calibrated predictions
    calibrated_pred = calibrator.predict_calibrated(image_batch)
    
    # Get top predictions
    top_idx = np.argmax(calibrated_pred[0])
    top_confidence = calibrated_pred[0][top_idx]
    
    # Determine if prediction is reliable
    is_reliable = (
        top_confidence >= confidence_threshold and 
        uncertainty[0] < 0.15  # Low epistemic uncertainty
    )
    
    # Generate Grad-CAM heatmap
    heatmap = gradcam.compute_heatmap(image_batch, class_idx=top_idx)
    overlaid_image = gradcam.overlay_heatmap(image_batch[0].astype(np.uint8), heatmap)
    
    # Top-3 predictions
    top3_idx = np.argsort(calibrated_pred[0])[-3:][::-1]
    top3_predictions = [
        (class_names[idx], float(calibrated_pred[0][idx]))
        for idx in top3_idx
    ]
    
    result = {
        'predicted_class': class_names[top_idx],
        'confidence': float(top_confidence),
        'uncertainty': float(uncertainty[0]),
        'entropy': float(entropy[0]),
        'is_reliable': is_reliable,
        'top3_predictions': top3_predictions,
        'heatmap': heatmap,
        'explanation_image': overlaid_image,
        'raw_probabilities': calibrated_pred[0].tolist()
    }
    
    return result

print("✅ Comprehensive prediction function ready")
```

```python
# ============================================================
# CELL 24: Visualize Predictions with Explanations
# ============================================================
def visualize_predictions(test_ds, model, calibrator, gradcam, class_names, num_samples=6):
    """Visualize predictions with Grad-CAM explanations"""
    fig, axes = plt.subplots(num_samples, 3, figsize=(15, num_samples * 4))
    
    sample_count = 0
    for images, labels in test_ds:
        for i in range(min(len(images), num_samples - sample_count)):
            image = images[i].numpy()
            true_label = int(labels[i].numpy())
            
            # Get prediction with explanation
            result = predict_with_explanation(
                model, image, calibrator, gradcam, 
                class_names, n_mc_samples=config.MC_DROPOUT_SAMPLES,
                confidence_threshold=config.CONFIDENCE_THRESHOLD
            )
            
            row = sample_count
            
            # Original image
            axes[row, 0].imshow(image.astype(np.uint8))
            axes[row, 0].set_title(f"True: {class_names[true_label]}")
            axes[row, 0].axis('off')
            
            # Grad-CAM heatmap
            axes[row, 1].imshow(result['heatmap'], cmap='jet')
            axes[row, 1].set_title('Grad-CAM Attention')
            axes[row, 1].axis('off')
            
            # Overlaid explanation
            axes[row, 2].imshow(result['explanation_image'])
            status = "✅" if result['is_reliable'] else "⚠️ Uncertain"
            title = f"Pred: {result['predicted_class']}\n"
            title += f"Conf: {result['confidence']:.1%} | Unc: {result['uncertainty']:.3f}\n"
            title += status
            axes[row, 2].set_title(title, fontsize=10)
            axes[row, 2].axis('off')
            
            sample_count += 1
            if sample_count >= num_samples:
                break
        
        if sample_count >= num_samples:
            break
    
    plt.tight_layout()
    plt.savefig(os.path.join(config.SAVED_MODELS_DIR, 'predictions_with_explanations.png'), dpi=150)
    plt.show()

visualize_predictions(test_ds, model, calibrator, gradcam, class_names)
```

```python
# ============================================================
# CELL 25: Evaluate Model on Test Set
# ============================================================
print("📊 Evaluating on test set...")

# Standard evaluation
test_loss, test_accuracy = model.evaluate(test_ds_prep, verbose=1)
print(f"\nTest Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")

# Detailed evaluation with uncertainty
all_predictions = []
all_labels = []
all_uncertainties = []

for images, labels in test_ds:
    mean_pred, uncertainty, _ = predict_with_uncertainty(
        model, images.numpy(), n_samples=config.MC_DROPOUT_SAMPLES
    )
    all_predictions.extend(np.argmax(mean_pred, axis=1))
    all_labels.extend(labels.numpy().astype(int))
    all_uncertainties.extend(uncertainty)

all_predictions = np.array(all_predictions)
all_labels = np.array(all_labels)
all_uncertainties = np.array(all_uncertainties)

# Accuracy for high-confidence predictions
confident_mask = all_uncertainties < 0.1
if confident_mask.sum() > 0:
    confident_accuracy = (all_predictions[confident_mask] == all_labels[confident_mask]).mean()
    print(f"\nAccuracy on confident predictions ({confident_mask.sum()}/{len(confident_mask)}): {confident_accuracy:.4f}")
```

```python
# ============================================================
# CELL 29: K-Fold Cross Validation (Optional)
# ============================================================
def run_kfold_validation(dataset, num_classes, class_names, k=5):
    """Run K-Fold cross validation for robust evaluation"""
    
    # Collect all data
    all_images = []
    all_labels = []
    for images, labels in dataset:
        all_images.append(images.numpy())
        all_labels.append(labels.numpy())
    
    all_images = np.concatenate(all_images, axis=0)
    all_labels = np.concatenate(all_labels, axis=0)
    
    print(f"Total samples: {len(all_labels)}")
    
    # Stratified K-Fold
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=42)
    
    fold_results = []
    
    for fold, (train_idx, val_idx) in enumerate(skf.split(all_images, all_labels)):
        print(f"\n{'='*50}")
        print(f"Fold {fold + 1}/{k}")
        print(f"{'='*50}")
        
        # Create fold datasets
        train_images, train_labels = all_images[train_idx], all_labels[train_idx]
        val_images, val_labels = all_images[val_idx], all_labels[val_idx]
        
        # Create tf.data.Dataset
        train_ds_fold = tf.data.Dataset.from_tensor_slices((train_images, train_labels))
        train_ds_fold = train_ds_fold.shuffle(1000).batch(config.BATCH_SIZE)
        train_ds_fold = train_ds_fold.map(apply_augmentation).prefetch(AUTOTUNE)
        
        val_ds_fold = tf.data.Dataset.from_tensor_slices((val_images, val_labels))
        val_ds_fold = val_ds_fold.batch(config.BATCH_SIZE)
        val_ds_fold = val_ds_fold.map(lambda x, y: (x, tf.cast(y, tf.float32))).prefetch(AUTOTUNE)
        
        # Build fresh model
        fold_model, fold_base = build_model(num_classes, config.BACKBONE, config.DROPOUT_RATE)
        fold_model.compile(
            optimizer=keras.optimizers.AdamW(learning_rate=1e-4, weight_decay=1e-5),
            loss=FocalLoss(gamma=2.0, alpha=0.25, num_classes=num_classes),
            metrics=[SoftAccuracy()]
        )
        
        # Train
        history = fold_model.fit(
            train_ds_fold,
            validation_data=val_ds_fold,
            epochs=15,
            callbacks=[EarlyStopping(patience=5, restore_best_weights=True)],
            verbose=1
        )
        
        # Evaluate
        val_loss, val_acc = fold_model.evaluate(val_ds_fold, verbose=0)
        fold_results.append({'fold': fold + 1, 'val_loss': val_loss, 'val_accuracy': val_acc})
        print(f"Fold {fold + 1} - Val Loss: {val_loss:.4f}, Val Accuracy: {val_acc:.4f}")
        
        # Clear memory
        del fold_model, fold_base
        tf.keras.backend.clear_session()
    
    # Summary
    print(f"\n{'='*50}")
    print("K-Fold Cross Validation Results")
    print(f"{'='*50}")
    
    accuracies = [r['val_accuracy'] for r in fold_results]
    print(f"Mean Accuracy: {np.mean(accuracies):.4f} ± {np.std(accuracies):.4f}")
    
    return fold_results

# Run K-Fold if enabled
if config.USE_KFOLD:
    kfold_results = run_kfold_validation(dataset, num_classes, class_names, k=config.K_FOLDS)
else:
    print("ℹ️ K-Fold validation skipped (set config.USE_KFOLD=True to enable)")
```
